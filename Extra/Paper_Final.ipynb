<H3> Abstract

This research explores the integration of on-chain data, staking data, and technical analysis as inputs
into a Multi-Layer Perceptron (MLP) model to classify buy, hold, and sell signals on an hourly time frame.
Pruned Exact Linear Time (PELT) segmentation algorithm is employed for change point detection, which
detect structural shifts in price trends since early 2021. The MLP classifier incorporates on-chain data,
staking data, and key technical indicators, including Exponential Moving Averages (EMA), Bollinger
Bands, and Fibonacci Retracement levels as input, and it’s designed to predict market movements across
different market phases. The model is trained using the Adam optimizer, evaluated through standard
performance metrics, and validated for robustness under repeated testing.

<H3> Introduction

Cryptocurrency has evolved immensely over the past decade. Once considered speculative, it has now
transformed into a recognized asset class for secure digital transactions. As the market expanded, so did the
methodologies for creating and securing digital assets.
Bitcoin, the first cryptocurrency, was built on the Proof of Work (PoW) system. This required miners
to solve complex mathematical problems to validate transactions, ensuring security through computational
power. While PoW was secure, it was also highly energy-intensive. Circa 2020, Proof of Stake (PoS) emerged
as a more energy-efficient alternative, and this method of network consensus has been the most popular.
Instead of mining, PoS selects validators based on their staked cryptocurrency, reducing energy consumption
while maintaining network security through economic incentives.
In our study, we utilize three types of features to predict price direction of various cryptocurrencies in
a portfolio: on-chain data, staking data, and technical indicators. We utilize this feature set to predict
one period ahead price movements using a 3-class MLP (Buy, Hold, Sell) for five cryptocurrencies; using
the softmax activation function, position sizes are computed as a percentage of prediction probabilities in a
dynamic portfolio.
In Section 2, we further explain the details of the data we’re using for the experiment. In Section 3,
we outline the implementation of the algorithm, as well as the explanation of the PELT Algorithm used
for change in point detection normalization in Section 3.1. For the rest of Section 3, we also introduce the
computation of technical indicators and methodology for signal (label) generation, explain our train-test-
split structure and data transformations, and introduce the Neural Network Architecture as well as show
how we evaluate the model and the respective performance metrics. In Section 4, we explain the portfolio
allocation methodology. In Section 5, we include an Analysis for each coin and on the portfolio. In Section
6, we include a conclusion where we also provide suggested future applied research.

This study analyzes the following cryptocurrencies:

On-chain blockchain data contains the information that is recorded directly on the blockchain ledger,
which includes details such as transaction volumes, the number of active wallet addresses, token transfers,
and processing difficulty, particularly for Proof of Work coins. However, our analysis is on Proof of Stake
coins, where we primarily have gas fees and transaction size per block as our main features. Since all trans-
actions on a blockchain are publicly accessible, this data provides a transparent and verifiable record of
network activity. For example, high transaction volumes and increasing active addresses can signal growing
user interest and network adoption, potentially leading to price appreciation. Additionally, shifts in token
distribution or movements into cold storage and staking contracts can reduce the circulating supply, creating
upward pressure on prices. Therefore, on-chain data provide insight into network health and stability, which

Staking data is a key component of our analysis, as it includes blockchain metrics such as the staking
ratio, total staked coins, total staking wallets, real reward rate, annualized rewards in USD, total active

flow of the blockchain for the selected cryptocurrencies. For example, percentage of eligible tokens that
are currently being staked or delegated to the network. This metric can provide insights into the level of

<H3> Implementation

This section walks through the steps of the trading strategy. The trading strategy follows these steps:
1. Collect and preprocess data, normalizing on-chain and technical indicator values.
2. Apply PELT segmentation to detect regime changes.
3. Train an MLP classifier using labeled buy, hold, and sell signals.
4. Evaluate performance using backtesting on hourly data.

<h3> Signal Generation

We compute the price difference, which is the arithmetic difference between consecutive closing prices.
Then, a threshold is set at half the standard deviation of the price difference. Based on the magnitude of
this threshold, the signal is categorized into:
• Buy: When the price difference is higher than the positive threshold.
• Sell: When the price difference is lower than the negative threshold.
• Hold: When the price difference is within the positive and negative threshold.

<H3> Nueral Network

A multilayer perceptron is a type of artificial neural network consisting of multiple layers of neurons.
The input layer receives the raw inputs from the columns of the data. The hidden layers process and
learn complex patterns in the data by applying an activation function to a weighted sum of its inputs.
The output layer produces the final probability prediction for buy, sell, and hold. Each unit within one
layer will produce an output, which will be the input to every unit of the next layer. The data will pass
through the MLP layer by layer, while being transformed at every unit, until it reaches the final output layer.
Figure 1: Diagram of a MLP architecture
For this particular neural network, the architecture has an input layer matching the number of features.
There are two hidden dense layers with 64 and 32 neurons, using LeakyReLU activation. Finally, there is a
single node in the output layer, using softmax activation.
5
The LeakyReLU activation function is similar to the standard ReLU activation function, except a small
positive gradient is set when the unit is inactive. If the gradient is 0 like the inactive ReLU unit, the learning
process will stop altogether at this unit, which prevents the model from reaching the optimal solution. This
is called the vanishing gradient problem, which is exactly what LeakyReLU mitigates.
figure 2: ReLU vs LeakyReLU
The softmax activation function transforms the raw inputs into a vector of probabilities, by taking into
account the relative magnitudes of each of the inputs. This is how the model expresses its belief for the
probabilities of the label that it’s trying to predict.
figure 3: Softmax Activation Function
The model is trained using the Adam optimizer, and trained for 50 epochs with a batch size of 32.

<H3> Portfolio Allocation

4 Portfolio Allocation
4.1 Position Sizing
Our portfolio allocation strategy takes advantage of the probabilities output by the softmax activation
function. At the last layer of the MLP network, the softmax activation function returns both a classification
prediction and associated probabilities for each of the possible classifications in the output space. For our
strategy we utilized a set initial capital of 100,000 USD with no leverage. For transaction cost we used 50
bps per trade.
In setting up the portfolio, we assume the equilibrium assumption and assign each token 20% of the
initial starting capital as its investment pool. When we receive a ’BUY’ signal, we take the corresponding
softmax probability associated with the ’BUY’ classification and treat it as a strength indicator. We size
the position by taking the strength indicator divided by some constant C to get some proportion p of our
maximum capital allocation which is used for the position sizing. This approach allows us to add varying
degrees of confidence to any given position, ensuring that our model doesn’t take identical actions on signals
of varying strength.
We perform a symmetrical procedure when we receive a ’SELL’ signal. Our team experimented with an
additional parameter for asymmetric sizing between ’BUY’ and ’SELL’ signals, which in essence would have
replaced our C parameter with CB , CS parameters. However, we found this extra layer of complexity diluted
the focus of the model and it’s ability to predict directional movements. Our team reworked the asymmetry
between ’BUY’ and ’SELL’ signals in our implementation of the stoploss parameter, which triggers more
excessive selling in times of portfolio stress.
’HOLD’ signals lead to no changes in the portfolio holdings.
4.2 Risk Control
A portion of our risk control is a result of our equilibrium assumption made in our position sizing. In
equalizing our maximum allocation to any given token, we introduce no Bayesian priors as to which token
is expected to perform best. Furthermore, we protect our portfolio from overexposure to any given token
should one experience a ’hot streak’ (generating consecutive ’BUY’ signals before experiencing a crash to
normal price levels).
Our risk control strategy also considers the importance of asymmetrical position actions. In an instance
such as the one described above, our strategy realizes the importance of closing out a position entirely. This
risk is controlled via our stoploss parameter, which is set to trigger when a token’s portfolio value drops by
.55% in a given time interval.

<h3> Conclusion

From this model, we demonstrated that on-chain data is valuable for designing trading strategies in
cryptocurrencies. We also demonstrated that MLP with LeakyReLU activations is effective for learning the
complexities in the on-chain and technical data.
We were able to train the MLP model in under 2 minutes for all the coins in our analysis, for approxi-
mately 4 years of hourly data. This shows that the model can be easily ported to a different asset, or for a
different time scale.
While there are difference in the training accuracy for different coins due to the inherent idiosyncrasies of
each individual coin, the training and testing accuracy for the same coin is highly stable. This suggests that
the MLP model itself is fairly robust from the stochastic optimization process. Additionally, the prediction
accuracy between the training and the test sets have been reasonably close, suggesting that the model is not
overfit.
Finally, there are many promising ways to improve the model, many of which can dramatically improve
the expressive power of the MLP model. This can potentially lead to significant improvements in strategy
performance

<h3> Drawbooks

Throughout the development of this strategy, we encountered several challenges that impacted data col-
lection, model accuracy, and overall predictive performance. Accessing reliable and cost-effective on-chain
and staking data for various cryptocurrencies proved to be a significant hurdle, requiring us to leverage
multiple third-party APIs such as DuneAPI, COINAPI, and StakingRewards. However, differences in data
structures across cryptocurrencies made it difficult to establish consistent patterns for prediction. Addi-
tionally, gaps and inconsistencies in certain datasets, such as NEAR and DOT, where missing values were
replaced with zeros, introduced errors that affected the model’s performance. These issues highlighted the
complexity of analyzing fundamentally different cryptocurrencies under a unified framework. Moving for-
ward, we aim to expand our dataset to include a broader universe of assets over a consistent time scale and
explore deeper relationships between PoS and PoW mechanisms to develop a more robust MLP model.
Since our strategy encompassed building an MLP model for each coin, we don’t reap the benefits of
stacking, which is the idea that we can train one model on all the data (using dummy variables to denote
the particular coin from the sample), which allows us to substantially increase the number of samples our
model can train with.
A deeper and wider MLP model will be able to capture more complexities in the structure of the data.
However, we chose not to explore architectures that were much bigger due to the significantly increased
model training time. However, with more computational resources, it’s possible that a larger model can
outperform our implementation.
The signal itself is labeled into three classes, buy, sell, hold. The model could only express its belief
about the probability between these three classes, but it cannot express any beliefs about the magnitude of
the predicted move. Moving away from a classification paradigm into a regression paradigm will give the
model even more flexibility.
If the trading strategy is to be implemented for real time trading, adding more data points can cause
the PELT segments to change retroactively under extreme market conditions. The model is designed to
perform under data that is normalized using these segments, but we do not have access to regime changes
on a walk-forward basis, and these regime changes may cause the model to struggle.
Also, an interesting edit to the structure of our train-test-split would be a 90% train, 10% test since we’ll
have more data to train with; due to our hourly data frequency, we’ll have a sufficient amount to perform
an out-of-sample backtest.
Another drawback is that we did not use leverage to fully exploit the strategy. With marginal funding
costs, this strategy may have potential for greater profits.
